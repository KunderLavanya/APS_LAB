## Birthday problem

```python
import matplotlib.pyplot as plt
import numpy as np

def bday_prob(sample):
    people_list = []
    probability_list = []
    loop_limit = min(sample, 365)
    prob_no_match = 1.0

    for i in range(1, loop_limit + 1):
        prob_no_match *= (365 - (i - 1)) / 365
        prob_match = 1 - prob_no_match
        people_list.append(i)
        probability_list.append(prob_match)
    if sample > 365:
        for i in range(366, sample + 1):
            people_list.append(i)
            probability_list.append(1.0)
    return people_list, probability_list
    njkh
size = 60
people, probabilities = bday_prob(size)

plt.figure(figsize=(8, 5))
plt.plot(people, probabilities, color='blue', linewidth=2, label="Probability")

plt.axhline(y=0.5, color='red', linestyle='--', alpha=0.7, label="50% Probability")
plt.axvline(x=23, color='green', linestyle=':', alpha=0.7, label="23 People")

plt.title("The Birthday Paradox: Probability of a Shared Birthday", fontsize=12)
plt.xlabel("Number of People in the Room", fontsize=10)
plt.ylabel("Probability of at Least One Match", fontsize=10)
plt.grid(True, alpha=0.3)
plt.legend()
plt.ylim(0, 1.05)
plt.show()
```
