```python
# Complete the exercise and write the learning in one of the cell

# Problem1: A fair coin is tossed 100 times.Calculate:

# Number of heads

# Number of tails
import random
total_tosses = 100
sides = ['Head', 'Tail']
results = [random.choice(sides) for _ in range(total_tosses)]
#count of heads and tails
no_heads = results.count('Head')
no_tails = results.count('Tail')

# Experimental probability of Head

# Experimental probability of Tail
p_head= no_heads / total_tosses
p_tail= no_tails / total_tosses

# 5. Output
print(f"Number of Heads: {no_heads}")
print(f"Number of Tails: {no_tails}")
print(f"Experimental Probability of Head: {p_head}")
print(f"Experimental Probability of Tail: {p_tail}")
```

```python
# Problem 2: check Effect of the number of trials

# Repeat the coin-toss experiment for: n=10, 100, 1,000, 10,000
import random

sides = ['Head', 'tails']

# Generate one sample of 10,000 tosses
results = [random.choice(sides) for _ in range(10000)]

sample_sizes = [10, 100, 1000, 10000]

for total_tosses in sample_sizes:

    sample = results[:total_tosses]

    no_heads = sample.count('Head')
    no_tails = sample.count('tails')

    p_head = no_heads / total_tosses
    p_tails = no_tails / total_tosses

    print(f"\nExperiment for sample size: {total_tosses}")
    print(f"Number of Heads: {no_heads}")
    print(f"Number of Tails: {no_tails}")
    print(f"Experimental Probability of Head: {p_head}")
    print(f"Experimental Probability of Tail: {p_tails}")
```

```python
# Rolling a die: A fair die is rolled 6,000 times.

# Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}
import random
from collections import Counter
total=6000
sample=[1,2,3,4,5,6]
results = [random.randint(1,6) for _ in range(total)]
freq=Counter(results)
# Print results
for outcome in sorted(freq.keys()):

    prob = freq[outcome] / total
    print(f"Outcome {outcome}:  Probability = {prob:.4f}")

```

```python
# Addition rule for disjoint events

# A die is rolled 10,000 times.

# Define: A=getting 1, B= getting 6, check the events are joint or disjoint

# Verify: P(A∪B)=P(A)+P(B)
import random
total=10000
choices=[random.randint(1,6) for _ in range(total)]
event_a=choices.count(1)
event_b=choices.count(6)

event_a_or_b = sum(1 for x in choices if x == 1 or x == 6)

p_a=event_a/total
p_b=event_b/total
p_a_union_b_measured = event_a_or_b / total

prob=p_a+p_b

# 4. Verify identity
print(f"Empirical P(A)         = {p_a:.4f}")
print(f"Empirical P(B)         = {p_b:.4f}")
print(f"Measured P(A∪B)        = {p_a_union_b_measured:.4f}")

print(f"Identity Verified?     = {p_a_union_b_measured == prob}")
```
